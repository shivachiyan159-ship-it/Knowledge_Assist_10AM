```python
os.makedirs(os.path.join(os.getcwd(),"logs"))
```

```python
from datetime import datetime
datetime.now()  # 11_05_2026.log
```

```python
LOG__FILE=f"{datetime.now().strftime('%m_%d_%Y')}.log"
LOG__FILE
```

```python
LOG_FOLDER=os.path.join(os.getcwd(),"logs")
LOG_FILE_PATH=os.path.join(LOG_FOLDER,LOG__FILE)
LOG_FILE_PATH
```

```python
# python onbuilt logging module
import logging
logging.basicConfig(
    filename=LOG_FILE_PATH,
    format="[ %(asctime)s ] %(levelname)s %(name)s (line:%(lineno)d) - %(message)s",
    level=logging.INFO
)
```
